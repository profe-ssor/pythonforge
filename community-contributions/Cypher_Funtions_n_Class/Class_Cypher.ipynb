Student Management System

```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course
        self.grades = []
    
    def add_grade(self, score):
        self.grades.append(score)
    
    def calculate_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)
    
    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Course: {self.course}")
        print(f"Average Grade: {self.calculate_average()}")

student1 = Student("Collins", 22, "Computer Science")
student1.add_grade(80)
student1.add_grade(90)
student1.add_grade(70)

student1.display_info()
```

Bank Account System

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposit Successful.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal Successful.")
        else:
            print("Insufficient Funds.")

    def check_balance(self):
        print(f"Current Balance: GHC {self.balance}")

    def account_details(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Account Number: {self.account_number}")
        print(f"Balance: GHC {self.balance}")

account = BankAccount("John Doe", "ACC001", 500)
account.deposit(300)
account.withdraw(100)

account.check_balance()
```

Library Book Management

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def borrow_book(self):
        if self.available:
            self.available = False
            print("Book borrowed successfully.")
        else:
            print("Book is not available.")

    def return_book(self):
        self.available = True
        print("Book returned successfully.")

    def book_status(self):
        if self.available:
            print("Status: Available")
        else:
            print("Status: Not Available")

book1 = Book("Python Programming", "Guido van Rossum")

book1.borrow_book()
book1.book_status() 
```

Employee Payroll System

```python
class Employee:
    def __init__(self, name, position, salary, tax_rate):
        self.name = name
        self.position = position
        self.salary = salary
        self.tax_rate = tax_rate

    def calculate_tax(self):
        return self.salary * self.tax_rate

    def net_salary(self):
        return self.salary - self.calculate_tax()

    def employee_details(self):
        print(f"Name: {self.name}")
        print(f"Position: {self.position}")
        print(f"Salary: GHC {self.salary}")
        print(f"Tax Rate: {self.tax_rate * 100}%")
        print(f"Net Salary: GHC {self.net_salary()}")


employee1 = Employee("Sarah", "Developer", 5000, 0.10)
print(employee1.net_salary())
```

Shopping Cart System

```python
class ShoppingCart:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = []

    def add_item(self, product, price):
        self.items.append((product, price))

    def remove_item(self, product):
        for item in self.items:
            if item[0] == product:
                self.items.remove(item)
                print(product, "removed successfully")      
                return
        print("Item not found")

    def calculate_total(self):
        total = 0
        for item in self.items:
            total += item[1]
        return total

    def display_cart(self):
        print("Customer:", self.customer_name)
        print("\nItems:")
        for item in self.items:
            print(item[0], "- GHC", item[1])

        print("\nTotal:", self.calculate_total())

cart = ShoppingCart("Collins")

cart.add_item("Laptop", 5000)
cart.add_item("Mouse", 100)

cart.display_cart()
```

Hospital Patient Management System

```python
class Patient:
    def __init__(self, name, age, disease):
        self.name = name
        self.age = age
        self.disease = disease
        self.admitted = False

    def admit_patient(self):
        self.admitted = True
        print(f"{self.name} has been admitted.")

    def discharge_patient(self):
        self.admitted = False
        print(f"{self.name} has been discharged.")

    def patient_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Disease: {self.disease}")

        if self.admitted:
            print("Status: Admitted")
        else:
            print("Status: Not Admitted")


patient1 = Patient("Michael", 30, "Malaria")
patient1.admit_patient()
patient1.patient_info()
```
