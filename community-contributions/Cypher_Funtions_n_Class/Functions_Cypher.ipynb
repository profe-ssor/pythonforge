Question 1 - Student Grade Function

```python
def calculate_grade(score):
    if 80 <= score <= 100:
        return 'A'
    elif 70 <= score <= 79:
        return 'B'
    elif 60 <= score <= 69:
        return 'C'
    elif 50 <= score <= 59:
        return 'D'
    else:
        return 'F'

scores = [85, 72, 48]
for score in scores:
    print(f'Score: {score}, Grade: {calculate_grade(score)}')
```

Question 2 - Shopping Cart Total

```python
def calculate_total(*prices):
    return sum(prices)

total = calculate_total(25, 40, 15, 20, 30)
print("Total Cost: GHC", total)
```

Question 3 - Employee Information

```python
def employee_details(**details):
    for key, value in details.items():
        print(f"{key.title()}: {value}")

employee_details(
    name = "John",
    department = "IT",
    salary = 5000,
    city = "Accra"
)
```

Question 4 - Bank withdrawal system

```python
def withdraw(balance, amount):
    if amount > balance:
        return "Insufficient funds"
    else:
        balance -= amount
        return balance

balance = 1000
print("New Balance:", withdraw(balance, 250))
print(withdraw(balance, 1500))
```

Question 5 - Temperature Converter

```python
def celcius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_to_celcius(fahrenheit):
    return (fahrenheit - 32) * 5/9

print("25°C =", celcius_to_fahrenheit(25), "°F")
print("98.6°F =", fahrenheit_to_celcius(98.6), "°C")
```

Question 6 - Student Management System

```python
students = []

def add_student(name):
    students.append(name)
    print(name, "added successfully.")

def remove_student(name):
    if name in students:
        students.remove(name)
        print(name, "removed successfully.")
    else:
        print(name, "not found.")

def display_students():
    print("Students:")
    for student in students:
        print(student)

def search_student(name):
    if name in students:
        print(name, "found.")
    else:
        print(name, "not found.")

add_student("Cypher")
add_student("Collins")
add_student("Rose")

display_students()
search_student("Cypher")
remove_student("Rose")
display_students()
```
